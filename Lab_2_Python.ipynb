*Ejercicio 3 - Map*

a. Dada una lista de temperaturas en Celsius, convertirlas a Fahrenheit.

```python
celsius = [0, 20, 37, 100]
fahrenheit = list(map(lambda c: (c * 9/5) + 32, celsius))
print(fahrenheit)
```

b. Dadas dos listas con la misma cantidad de elementos, generar una nueva sumando cada elementos (primer elemento de lista 1 + primer elemento lista 2).

```python
lista1 = [1, 2, 3, 4]
lista2 = [10, 20, 30, 40]
suma_listas = list(map(lambda x, y: x + y, lista1, lista2))
print(suma_listas)
```

c. Dada una lista de radios de circulos, generar la lista de sus areas.

```python
import math

radios = [1, 2.5, 5, 10]
areas = list(map(lambda r: math.pi * (r ** 2), radios))
print(areas)
```

d. Dada un diccionario, generar una lista de tuplas (nombre, valor+10%)

```python
precios = {"pan": 1000, "leche": 2500, "café": 5000}
resultado = list(map(lambda item: (item[0], item[1] * 1.10), precios.items()))
print(resultado)
```

e. Dada una matriz, multiplicar todos los elementos por 10.

```python
matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
matriz_x10 = list(map(lambda fila: list(map(lambda x: x * 10, fila)), matriz))
print(matriz_x10)
```

Ejercicio 4 - Filter

a. palabras con mas de 4 letras

```python
palabras = ["sol", "ventana", "luz", "computadora", "mar"]

resultado = list(filter(lambda p: len(p) > 4, palabras))
print(resultado)  # ['ventana', 'computadora']
```

b. filtrar elementos que no son nulos

```python
elementos = ["a", None, 42, None, True, "b"]

no_nulos = list(filter(lambda x: x is not None, elementos))
print(no_nulos)  # ['a', 42, True, 'b']
```

c. palabras que empiezan por una vocal

```python
palabras = ["arbol", "casa", "elefante", "iglesia", "oso", "uva", "perro"]
vocales = ("a", "e", "i", "o", "u", "á", "é", "í", "ó", "ú")

resultado = list(filter(lambda p: p.lower().startswith(vocales), palabras))
print(resultado)  # ['arbol', 'elefante', 'iglesia', 'oso', 'uva']
```

d.palabras que son palindromos

```python
palabras = ["oso", "python", "reconocer", "radar", "codigo", "ana"]

resultado = list(filter(lambda p: p.lower() == p.lower()[::-1], palabras))
print(resultado)  # ['oso', 'reconocer', 'radar', 'ana']
```

e. numeros que terminan en 5

```python
numeros = [5, 12, 15, 23, 35, 100, -25, 7]

resultado = list(filter(lambda n: abs(n) % 10 == 5, numeros))
print(resultado)  # [5, 15, 35, -25]
```

f. productos con precio mayor a 100

```python
productos = [
    {"nombre": "Teclado", "precio": 80},
    {"nombre": "Mouse", "precio": 40},
    {"nombre": "Monitor", "precio": 300},
    {"nombre": "Webcam", "precio": 150},
]

caros = list(filter(lambda prod: prod["precio"] > 100, productos))
print(caros)
```

EJERCICIO 4 REDUCE

a. multiplicacion de una lista de numeros

```python
from functools import reduce

numeros = [2, 3, 4, 5]

producto = reduce(lambda acc, x: acc * x, numeros)
print(producto)  # 120
```

b. concatenar palabras en una sola cadena

```python
from functools import reduce

palabras = ["Hola", "mundo", "desde", "Python"]

# Concatenación directa:
union_directa = reduce(lambda acc, p: acc + p, palabras)
print(union_directa)  # "HolamundodesdePython"

# Concatenación separada por espacios:
con_espacios = reduce(lambda acc, p: f"{acc} {p}", palabras)
print(con_espacios)  # "Hola mundo desde Python"
```

c. hallar el mayor numero con una funcion auxiliar

```python
from functools import reduce


def mayor_de_dos(a, b):
    return a if a > b else b


numeros = [14, 67, 3, 89, 42, 89, 19]

mayor = reduce(mayor_de_dos, numeros)
print(mayor)  # 89
```
